```python
import numpy as np
import matplotlib.pyplot as plt

# 乱数シード（再現性用）
np.random.seed(0)

# サンプル数と分布パラメータ
n_samples = 400
mu = 0.4054
sigma = 0.1411

# 0〜1に収まる正規分布風データ生成
samples = []
while len(samples) < n_samples:
    x = np.random.normal(mu, sigma)
    if 0.0 <= x <= 1.0:
        samples.append(x)

samples = np.array(samples)

# プロット
plt.figure()
plt.hist(samples, bins=30, density=True)
plt.xlabel("value")
plt.ylabel("density")
plt.title("Normal-like distribution in [0, 1] (400 samples)")
plt.xlim(0, 1)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm

# ===== CSVの絶対パス =====
csv_path = "/data2/ssk/githubtest/sim2real/data/fep/fep1mel_435_processed.csv"   # ←書き換え

# ===== 使用するカラム名 =====
column_name = "ddg_scaled01"                     # ←書き換え

# ===== CSV読み込み =====
df = pd.read_csv(csv_path)
data = df[column_name].values

# 念のため 0-1 の範囲に制限
data = data[(data >= 0.0) & (data <= 1.0)]

# ===== 正規分布パラメータ推定 =====
mu = np.mean(data)
sigma = np.std(data, ddof=0)

print(f"estimated mu = {mu:.4f}")
print(f"estimated sigma = {sigma:.4f}")

# ===== 正規分布 =====
x = np.linspace(0, 1, 500)
y = norm.pdf(x, mu, sigma)

# ===== プロット =====
plt.figure()
plt.hist(data, bins=30, density=True, alpha=0.6, label="data")
plt.plot(x, y, label="normal approximation")
plt.xlim(0, 1)
plt.xlabel("value")
plt.ylabel("density")
plt.legend()
plt.title("Normal approximation of (0-1) data")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ===== 設定 =====
csv_path_1 = "/data2/ssk/githubtest/sim2real/data/fep/fep4idl_409_processed.csv"
csv_path_2 = "/data2/ssk/githubtest/sim2real/data/mpnn/4idlMPNN2_processed.csv"
column_name = "ddg_scaled01"
bins = 30
# =================

# CSV読み込み
df1 = pd.read_csv(csv_path_1)
df2 = pd.read_csv(csv_path_2)

# ddgのみ取り出して結合（行対応が前提）
merged = pd.concat(
    [df1[column_name], df2[column_name]],
    axis=1,
    keys=["FEP", "thermoMPNN"]
).dropna()

data1 = merged["FEP"]
data2 = merged["thermoMPNN"]

# 相関係数（Pearson）
corr = data1.corr(data2)

# =================
# プロット
# =================
plt.figure(figsize=(10, 4))

# ---- ヒストグラム ----
plt.subplot(1, 2, 1)
plt.hist(data1, bins=bins, alpha=0.6, label="FEP")
plt.hist(data2, bins=bins, alpha=0.6, label="thermoMPNN")
plt.xlabel(column_name)
plt.ylabel("Frequency")
plt.title("Distribution_4idl")
plt.legend()

# ---- 相関（散布図） ----
plt.subplot(1, 2, 2)
plt.scatter(data1, data2, alpha=0.6)
plt.xlabel("FEP ddg")
plt.ylabel("thermoMPNN ddg")
plt.title(f"Pearson (r = {corr:.3f})_4idl")

plt.tight_layout()
plt.show()
```
