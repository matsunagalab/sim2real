```python
import pandas as pd

input_path = "/data2/ssk/githubtest/sim2real/data/mpnn/4idlMPNN2_processed.csv"
output_path = "/data2/ssk/githubtest/sim2real/data/mpnn/4idlMPNN2_processed_rand.csv"

df = pd.read_csv(input_path)

cols = ["ddg", "ddg_neg", "ddg_scaled01"]  # シャッフルしたい列セット

shuf = df[cols].sample(frac=1, random_state=None).reset_index(drop=True)
df_rand = df.copy()
df_rand[cols] = shuf

df_rand.to_csv(output_path, index=False)
```
