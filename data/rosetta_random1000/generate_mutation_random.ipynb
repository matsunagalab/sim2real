```python
import json
import os
import random
import re
import time
import pandas as pd
```

```python
# Replace with your antibody wild-type (WT) sequence
seq_name = "1mel"
wt_sequence = "VQLQASGGGSVQAGGSLRLSCAASGYTIGPYCMGWFRQAPGKEREGVAAINMGGGITYYADSVKGRFTISQDNAKNTVYLLMNSLEPEDTAIYYCAADSTIYASYYECGHGLSTGGYGYDSWGQGTQVTVSS"

num_sequences_to_generate = 1000
num_mutations = 2  # number of mutations per variant
output_dir = "."  # current directory (where the notebook is)
output_csv_filename = f"multi_mutations_random_{num_mutations}mut_{seq_name}_{num_sequences_to_generate}.csv"
```

```python
# Clean the input sequence
amino_acids = "ACDEFGHIKLMNPQRSTVWY"
mutant_amino_acids = "ADEFGHIKLMNPQRSTVWY"  # exclude Cys
wt_cleaned = "".join(re.findall(f"[{amino_acids}]", wt_sequence.upper()))
print(f"Cleaned WT sequence: {wt_cleaned[:30]}... (Length: {len(wt_cleaned)})")
print(f"Mutation target amino acids (excluding Cys): {mutant_amino_acids}")

print(f"\nGenerating {num_sequences_to_generate} mutant sequences with {num_mutations} mutations (random strategy)...")

random_seed = 42
random.seed(random_seed)
print(f"Random seed: {random_seed}")

all_mutation_data = []
generated_sequences = set()
start_time = time.time()
last_reported_count = 0

while len(all_mutation_data) < num_sequences_to_generate:
    if num_mutations > len(wt_cleaned):
        print(f"Error: num_mutations ({num_mutations}) exceeds sequence length ({len(wt_cleaned)})")
        break
    
    mutation_positions = random.sample(range(len(wt_cleaned)), num_mutations)
    mutations_applied = []
    new_sequence_list = list(wt_cleaned)
    
    for pos in mutation_positions:
        original_aa = wt_cleaned[pos]
        available_aas = [aa for aa in mutant_amino_acids if aa != original_aa]
        if len(available_aas) > 0:
            new_aa = random.choice(available_aas)
            mutations_applied.append({
                "position": pos + 1,  # 1-based
                "original_aa": original_aa,
                "mutated_aa": new_aa
            })
            new_sequence_list[pos] = new_aa
    
    if len(mutations_applied) == num_mutations:
        new_mutated_sequence = "".join(new_sequence_list)
        if new_mutated_sequence not in generated_sequences:
            generated_sequences.add(new_mutated_sequence)
            all_mutation_data.append({
                "sequence": new_mutated_sequence,
                "mutations": mutations_applied,
                "num_mutations": len(mutations_applied)
            })
            current_count = len(all_mutation_data)
            if current_count > last_reported_count and current_count % 10 == 0:
                elapsed = time.time() - start_time
                print(f"Generated {current_count} / {num_sequences_to_generate} sequences... ({elapsed:.2f}s)")
                last_reported_count = current_count

total_time = time.time() - start_time
print(f"\nSequence generation complete. Total time: {total_time:.2f}s")
print(f"Generated {len(all_mutation_data)} unique sequences with exactly {num_mutations} mutations.")

# Build DataFrame and save CSV
df_data = [{
    "sequence": d["sequence"],
    "num_mutations": d["num_mutations"],
    "mutations": json.dumps(d["mutations"])
} for d in all_mutation_data]
df_output = pd.DataFrame(df_data)
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, output_csv_filename)
df_output.to_csv(output_path, index=False)

print(f"Saved {len(df_output)} sequences to '{output_path}'")
print(f"Columns: {list(df_output.columns)}")
print(df_output.head())
```
